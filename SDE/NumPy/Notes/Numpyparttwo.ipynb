small **runnable example** with output.

> actually use 95% of the time, grouped so you can find them fast. Run any cell to see it work.

## 1. Creating Arrays
Functions that build new arrays from scratch, from lists, or from ranges.

### `np.array`

Create an ndarray from a list, tuple, or another array.

**Syntax:** `np.array(object, dtype=None, copy=True, ndmin=0)`

````
import numpy as np

arr = np.array([1, 2, 3], dtype=float)
print(arr)
````

### `np.asarray`

Convert input to an array. Unlike array(), it does NOT copy if input is already an ndarray.

**Syntax:** `np.asarray(a, dtype=None)`

````
import numpy as np

print(np.asarray([1, 2, 3]))
````

### `np.zeros`

Return a new array filled with 0s of the given shape.

**Syntax:** `np.zeros(shape, dtype=float)`

````
import numpy as np

print(np.zeros((2, 3)))
````

### `np.ones`

Return a new array filled with 1s of the given shape.

**Syntax:** `np.ones(shape, dtype=float)`

````
import numpy as np

print(np.ones((2, 2), dtype=int))
````

### `np.full`

Return a new array of given shape filled with a value you choose.

**Syntax:** `np.full(shape, fill_value, dtype=None)`

````
import numpy as np

print(np.full((2, 2), 7))
````

### `np.empty`

Return an array without initializing values (contents are garbage, fast to allocate).

**Syntax:** `np.empty(shape, dtype=float)`

````
import numpy as np

a = np.empty((2, 2))
print(a.shape)
````

### `np.zeros_like`

Return an array of zeros with the same shape & type as a given array.

**Syntax:** `np.zeros_like(a, dtype=None)`

````
import numpy as np

x = np.array([[1, 2], [3, 4]])
print(np.zeros_like(x))
````

### `np.ones_like`

Return an array of ones with the same shape & type as a given array.

**Syntax:** `np.ones_like(a, dtype=None)`

````
import numpy as np

x = np.array([[1, 2], [3, 4]])
print(np.ones_like(x))
````

### `np.full_like`

Return an array with the same shape as a, filled with a chosen value.

**Syntax:** `np.full_like(a, fill_value, dtype=None)`

````
import numpy as np

x = np.array([1, 2, 3])
print(np.full_like(x, 9))
````

### `np.arange`

Return evenly spaced values within a range (like Python range but returns an array).

**Syntax:** `np.arange(start, stop, step, dtype=None)`

````
import numpy as np

print(np.arange(0, 10, 2))
````

### `np.linspace`

Return 'num' evenly spaced numbers between start and stop (inclusive).

**Syntax:** `np.linspace(start, stop, num=50, endpoint=True)`

````
import numpy as np

print(np.linspace(0, 1, 5))
````

### `np.logspace`

Return numbers evenly spaced on a log scale.

**Syntax:** `np.logspace(start, stop, num=50, base=10.0)`

````
import numpy as np

print(np.logspace(0, 2, 3))
````

### `np.eye`

Return a 2-D identity-like array with 1s on a diagonal (k shifts the diagonal).

**Syntax:** `np.eye(N, M=None, k=0, dtype=float)`

````
import numpy as np

print(np.eye(3))
````

### `np.identity`

Return a square n x n identity matrix.

**Syntax:** `np.identity(n, dtype=float)`

````
import numpy as np

print(np.identity(2))
````

### `np.diag`

Extract a diagonal, or construct a diagonal array from a 1-D input.

**Syntax:** `np.diag(v, k=0)`

````
import numpy as np

print(np.diag([1, 2, 3]))
````

### `np.meshgrid`

Return coordinate matrices from coordinate vectors (useful for grids/plots).

**Syntax:** `np.meshgrid(*xi, indexing='xy')`

````
import numpy as np

x, y = np.meshgrid([1, 2, 3], [4, 5])
print(x)
print(y)
````

### `np.fromiter`

Create a 1-D array from an iterable object.

**Syntax:** `np.fromiter(iterable, dtype, count=-1)`

````
import numpy as np

print(np.fromiter((i*i for i in range(5)), dtype=int))
````

## 2. Inspecting Arrays (attributes)
These are *attributes*, not functions — you access them without parentheses.

````
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
print("shape   :", arr.shape)     # (rows, columns)
print("ndim    :", arr.ndim)      # number of dimensions
print("size    :", arr.size)      # total number of elements
print("dtype   :", arr.dtype)     # data type
print("itemsize:", arr.itemsize)  # bytes per element
print("T (transpose):\n", arr.T)
````

## 3. Reshaping & Manipulating Arrays
Change the shape, flatten, join, split, or repeat arrays.

### `np.reshape`

Give an array a new shape without changing its data. Use -1 to auto-calc a dimension.

**Syntax:** `np.reshape(a, newshape)`

````
import numpy as np

a = np.arange(6)
print(a.reshape(2, 3))
````

### `np.ravel`

Flatten an array to 1-D (returns a view when possible).

**Syntax:** `np.ravel(a)`

````
import numpy as np

a = np.array([[1, 2], [3, 4]])
print(np.ravel(a))
````

### `np.flatten`

Flatten to 1-D — always returns a COPY (method on the array).

**Syntax:** `ndarray.flatten(order='C')`

````
import numpy as np

a = np.array([[1, 2], [3, 4]])
print(a.flatten())
````

### `np.transpose`

Permute the dimensions (swap rows and columns for 2-D).

**Syntax:** `np.transpose(a, axes=None)`

````
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(np.transpose(a))
````

### `np.swapaxes`

Interchange two axes of an array.

**Syntax:** `np.swapaxes(a, axis1, axis2)`

````
import numpy as np

a = np.array([[1, 2], [3, 4]])
print(np.swapaxes(a, 0, 1))
````

### `np.expand_dims`

Insert a new axis (increase dimensions by one).

**Syntax:** `np.expand_dims(a, axis)`

````
import numpy as np

a = np.array([1, 2, 3])
print(np.expand_dims(a, axis=0).shape)
````

### `np.squeeze`

Remove axes of length 1 from the shape.

**Syntax:** `np.squeeze(a, axis=None)`

````
import numpy as np

a = np.array([[[1], [2], [3]]])
print(np.squeeze(a).shape)
````

### `np.concatenate`

Join a sequence of arrays along an existing axis.

**Syntax:** `np.concatenate((a1, a2, ...), axis=0)`

````
import numpy as np

a = np.array([1, 2]); b = np.array([3, 4])
print(np.concatenate((a, b)))
````

### `np.stack`

Join arrays along a NEW axis.

**Syntax:** `np.stack(arrays, axis=0)`

````
import numpy as np

a = np.array([1, 2]); b = np.array([3, 4])
print(np.stack((a, b), axis=0))
````

### `np.vstack`

Stack arrays vertically (row-wise).

**Syntax:** `np.vstack(tup)`

````
import numpy as np

a = np.array([1, 2, 3]); b = np.array([4, 5, 6])
print(np.vstack((a, b)))
````

### `np.hstack`

Stack arrays horizontally (column-wise).

**Syntax:** `np.hstack(tup)`

````
import numpy as np

a = np.array([1, 2, 3]); b = np.array([4, 5, 6])
print(np.hstack((a, b)))
````

### `np.split`

Split an array into multiple equal sub-arrays.

**Syntax:** `np.split(ary, indices_or_sections, axis=0)`

````
import numpy as np

a = np.arange(6)
print(np.split(a, 3))
````

### `np.hsplit`

Split an array horizontally (column-wise).

**Syntax:** `np.hsplit(ary, indices_or_sections)`

````
import numpy as np

a = np.arange(6).reshape(2, 3)
print(np.hsplit(a, 3))
````

### `np.append`

Append values to the end of an array (returns a new array).

**Syntax:** `np.append(arr, values, axis=None)`

````
import numpy as np

a = np.array([1, 2, 3])
print(np.append(a, [4, 5]))
````

### `np.insert`

Insert values at a given index.

**Syntax:** `np.insert(arr, obj, values, axis=None)`

````
import numpy as np

a = np.array([1, 2, 4])
print(np.insert(a, 2, 3))
````

### `np.delete`

Return a new array with elements at given index removed.

**Syntax:** `np.delete(arr, obj, axis=None)`

````
import numpy as np

a = np.array([1, 2, 3, 4])
print(np.delete(a, 1))
````

### `np.resize`

Return a new array of the given shape, repeating data if needed.

**Syntax:** `np.resize(a, new_shape)`

````
import numpy as np

a = np.array([1, 2, 3])
print(np.resize(a, (2, 3)))
````

### `np.tile`

Repeat a whole array a given number of times.

**Syntax:** `np.tile(A, reps)`

````
import numpy as np

a = np.array([1, 2])
print(np.tile(a, 3))
````

### `np.repeat`

Repeat each element of an array.

**Syntax:** `np.repeat(a, repeats, axis=None)`

````
import numpy as np

a = np.array([1, 2, 3])
print(np.repeat(a, 2))
````

### `np.flip`

Reverse the order of elements along an axis.

**Syntax:** `np.flip(m, axis=None)`

````
import numpy as np

a = np.array([1, 2, 3])
print(np.flip(a))
````

### `np.roll`

Shift elements along an axis (wraps around).

**Syntax:** `np.roll(a, shift, axis=None)`

````
import numpy as np

a = np.array([1, 2, 3, 4])
print(np.roll(a, 1))
````

### `np.unique`

Find the sorted unique elements of an array.

**Syntax:** `np.unique(ar, return_counts=False)`

````
import numpy as np

a = np.array([1, 2, 2, 3, 3, 3])
print(np.unique(a, return_counts=True))
````

### `np.multiply`

Element-wise multiplication.

**Syntax:** `np.multiply(x1, x2)`

````
import numpy as np

print(np.multiply([1, 2, 3], [4, 5, 6]))
````

### `np.divide`

Element-wise division.

**Syntax:** `np.divide(x1, x2)`

````
import numpy as np

print(np.divide([10, 20, 30], [2, 4, 5]))
````

### `np.power`

### `np.mod`

Element-wise remainder (modulo).

### `np.sqrt`

Element-wise square root.

### `np.exp`

Element-wise exponential (e**x).

### `np.log`

Natural logarithm (base e). Also np.log2, np.log10.

### `np.abs`

Element-wise absolute value.

### `np.round`

Round to the given number of decimals.

### `np.floor`

Round down to the nearest integer.

Round up to the nearest integer.

### `np.cumsum`

Cumulative sum of elements.

Cumulative product of elements.

**Syntax:** `np.cumprod(a, axis=None)`

````
import numpy as np

print(np.cumprod([1, 2, 3, 4]))
````

Element-wise sine (x in radians). Also cos, tan, arcsin, etc.

### `np.deg2rad`

````
import numpy as np

print(np.deg2rad([0, 90, 180]))
````

Summaries across a whole array or along an axis.

### `np.prod`

Product of array elements.

````
import numpy as np

print(np.prod([1, 2, 3, 4]))
````

### `np.min`

Minimum value. (np.max for maximum.)

### `np.mean`

Arithmetic mean (average).

**Syntax:** `np.mean(a, axis=None)`

````
import numpy as np

print(np.mean([1, 2, 3, 4]))
````

### `np.median`

Median (middle value).

**Syntax:** `np.median(a, axis=None)`

````
import numpy as np

print(np.median([1, 2, 3, 4, 5]))
````

### `np.average`

Weighted average.

### `np.std`

Standard deviation (spread of data).

### `np.var`

### `np.percentile`

### `np.ptp`

**Syntax:** `np.ptp(a, axis=None)`

````
import numpy as np

print(np.ptp([1, 5, 3]))
````

### `np.argmin`

Index of the minimum value. (np.argmax for maximum.)

### `np.corrcoef`

Pearson correlation coefficient matrix.

## 6. Logic, Comparison & Conditions
Boolean tests and conditional selection.

### `np.where`

Return elements from x where condition is True, else from y. With one arg, returns indices.

**Syntax:** `np.where(condition, x, y)`

````
import numpy as np

a = np.array([1, 2, 3, 4])
print(np.where(a > 2, a, 0))
````

### `np.all`

Test whether ALL elements are True.

**Syntax:** `np.all(a, axis=None)`

````
import numpy as np

print(np.all([True, True, False]))
````

### `np.any`

Test whether ANY element is True.

**Syntax:** `np.any(a, axis=None)`

````
import numpy as np

print(np.any([False, False, True]))
````

### `np.greater`

Element-wise x1 > x2. (Also less, equal, not_equal.)

**Syntax:** `np.greater(x1, x2)`

````
import numpy as np

print(np.greater([1, 5, 3], [2, 2, 2]))
````

### `np.logical_and`

Element-wise logical AND. (Also logical_or, logical_not.)

**Syntax:** `np.logical_and(x1, x2)`

````
import numpy as np

print(np.logical_and([True, False], [True, True]))
````

### `np.isnan`

Test element-wise for NaN (Not a Number).

**Syntax:** `np.isnan(x)`

````
import numpy as np

print(np.isnan([1.0, np.nan, 3.0]))
````

### `np.clip`

Limit values to a range: anything below a_min becomes a_min, above a_max becomes a_max.

**Syntax:** `np.clip(a, a_min, a_max)`

````
import numpy as np

print(np.clip([1, 5, 10, -2], 0, 6))
````

## 7. Sorting & Searching

### `np.sort`

Return a sorted copy of an array.

**Syntax:** `np.sort(a, axis=-1, kind='quicksort')`

````
import numpy as np

print(np.sort([3, 1, 2]))
````

### `np.argsort`

Return the indices that would sort an array.

**Syntax:** `np.argsort(a, axis=-1)`

````
import numpy as np

print(np.argsort([3, 1, 2]))
````

### `np.searchsorted`

Find indices where elements should be inserted to keep order.

**Syntax:** `np.searchsorted(a, v, side='left')`

````
import numpy as np

print(np.searchsorted([1, 2, 4, 5], 3))
````

### `np.nonzero`

Return the indices of non-zero elements.

**Syntax:** `np.nonzero(a)`

````
import numpy as np

print(np.nonzero([0, 3, 0, 5]))
````

### `np.count_nonzero`

Count the number of non-zero values.

**Syntax:** `np.count_nonzero(a, axis=None)`

````
import numpy as np

print(np.count_nonzero([0, 1, 2, 0, 3]))
````

### `np.extract`

Return elements that satisfy a condition.

**Syntax:** `np.extract(condition, arr)`

````
import numpy as np

a = np.array([1, 2, 3, 4])
print(np.extract(a > 2, a))
````

## 8. Set Operations
Treat arrays like mathematical sets.

### `np.intersect1d`

Return the sorted, unique values that are in BOTH arrays.

**Syntax:** `np.intersect1d(ar1, ar2)`

````
import numpy as np

print(np.intersect1d([1, 2, 3], [2, 3, 4]))
````

### `np.union1d`

Return the sorted, unique values from EITHER array.

**Syntax:** `np.union1d(ar1, ar2)`

````
import numpy as np

print(np.union1d([1, 2], [2, 3]))
````

### `np.setdiff1d`

Values in ar1 that are NOT in ar2.

**Syntax:** `np.setdiff1d(ar1, ar2)`

````
import numpy as np

print(np.setdiff1d([1, 2, 3], [2]))
````

### `np.in1d`

Test whether each element of ar1 is also in ar2.

**Syntax:** `np.in1d(ar1, ar2)`

````
import numpy as np

print(np.in1d([1, 2, 3], [2, 3]))
````

## 9. Linear Algebra
Matrix math. Many live under the `np.linalg` sub-module.

### `np.dot`

Dot product of two arrays (matrix multiplication for 2-D).

**Syntax:** `np.dot(a, b)`

````
import numpy as np

a = np.array([[1, 2], [3, 4]]); b = np.array([[5, 6], [7, 8]])
print(np.dot(a, b))
````

### `np.matmul`

Matrix product of two arrays.

**Syntax:** `np.matmul(a, b)  # or a @ b`

````
import numpy as np

a = np.array([[1, 0], [0, 1]]); b = np.array([[4, 1], [2, 2]])
print(a @ b)
````

### `np.inner`

Inner product of two arrays.

**Syntax:** `np.inner(a, b)`

````
import numpy as np

print(np.inner([1, 2, 3], [4, 5, 6]))
````

### `np.cross`

Cross product of two 3-element vectors.

**Syntax:** `np.cross(a, b)`

````
import numpy as np

print(np.cross([1, 0, 0], [0, 1, 0]))
````

### `np.linalg.inv`

Inverse of a square matrix.

**Syntax:** `np.linalg.inv(a)`

````
import numpy as np

a = np.array([[1, 2], [3, 4]])
print(np.linalg.inv(a))
````

### `np.linalg.det`

Determinant of a square matrix.

**Syntax:** `np.linalg.det(a)`

````
import numpy as np

a = np.array([[1, 2], [3, 4]])
print(round(np.linalg.det(a), 2))
````

### `np.linalg.eig`

Eigenvalues and eigenvectors of a square matrix.

**Syntax:** `np.linalg.eig(a)`

````
import numpy as np

a = np.array([[2, 0], [0, 3]])
vals, vecs = np.linalg.eig(a)
print(vals)
````

### `np.linalg.solve`

Solve a linear system a·x = b for x.

**Syntax:** `np.linalg.solve(a, b)`

````
import numpy as np

a = np.array([[3, 1], [1, 2]]); b = np.array([9, 8])
print(np.linalg.solve(a, b))
````

### `np.linalg.norm`

Vector or matrix norm (length/magnitude).

**Syntax:** `np.linalg.norm(x, ord=None)`

````
import numpy as np

print(np.linalg.norm([3, 4]))
````

## 10. Random Numbers
From the `np.random` module. `seed()` makes results reproducible.

### `np.random.seed`

Set the random seed so results are reproducible.

**Syntax:** `np.random.seed(n)`

````
import numpy as np

np.random.seed(0)
print(np.random.randint(100))
````

### `np.random.rand`

Random floats in [0, 1) of the given shape.

**Syntax:** `np.random.rand(d0, d1, ...)`

````
import numpy as np

np.random.seed(1)
print(np.random.rand(3))
````

### `np.random.randint`

Random integers from low (inclusive) to high (exclusive).

**Syntax:** `np.random.randint(low, high=None, size=None)`

````
import numpy as np

np.random.seed(2)
print(np.random.randint(1, 10, size=5))
````

### `np.random.randn`

Random floats from the standard normal distribution.

**Syntax:** `np.random.randn(d0, d1, ...)`

````
import numpy as np

np.random.seed(3)
print(np.round(np.random.randn(3), 2))
````

### `np.random.choice`

Random sample(s) from a 1-D array.

**Syntax:** `np.random.choice(a, size=None, replace=True, p=None)`

````
import numpy as np

np.random.seed(4)
print(np.random.choice([10, 20, 30], size=4))
````

### `np.random.uniform`

Random floats from a uniform distribution over [low, high).

**Syntax:** `np.random.uniform(low=0.0, high=1.0, size=None)`

````
import numpy as np

np.random.seed(5)
print(np.round(np.random.uniform(1, 5, 3), 2))
````

### `np.random.normal`

Random samples from a normal (Gaussian) distribution.

**Syntax:** `np.random.normal(loc=0.0, scale=1.0, size=None)`

````
import numpy as np

np.random.seed(6)
print(np.round(np.random.normal(0, 1, 3), 2))
````

### `np.random.shuffle`

Shuffle an array IN PLACE (modifies the original).

**Syntax:** `np.random.shuffle(x)`

````
import numpy as np

np.random.seed(7)
a = np.array([1, 2, 3, 4, 5])
np.random.shuffle(a)
print(a)
````

### `np.random.permutation`

Return a shuffled COPY (original stays unchanged).

**Syntax:** `np.random.permutation(x)`

````
import numpy as np

np.random.seed(8)
print(np.random.permutation([1, 2, 3, 4, 5]))
````

## 11. Saving & Loading Arrays
Persist arrays to disk and read them back.

### `np.save`

Save one array to a binary .npy file.

**Syntax:** `np.save(file, arr)`

````
import numpy as np

a = np.array([1, 2, 3])
np.save('/tmp/myarr.npy', a)
print('saved')
````

### `np.load`

Load an array back from a .npy file.

**Syntax:** `np.load(file)`

````
import numpy as np

print(np.load('/tmp/myarr.npy'))
````

### `np.savetxt`

Save an array to a text/CSV file.

**Syntax:** `np.savetxt(fname, X, delimiter=' ')`

````
import numpy as np

a = np.array([[1, 2], [3, 4]])
np.savetxt('/tmp/data.csv', a, delimiter=',')
print('saved')
````

### `np.loadtxt`

Load data from a text/CSV file.

**Syntax:** `np.loadtxt(fname, delimiter=None)`

````
import numpy as np

print(np.loadtxt('/tmp/data.csv', delimiter=','))
````
