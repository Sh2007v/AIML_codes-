# Matplotlib Function Reference — A Beginner's Guide

**Matplotlib** is Python's most popular library for making charts and graphs. Its main
module is `pyplot`, imported as `plt`. This notebook is a categorized reference of the
functions you'll use most — each with a description, its **syntax with parameters**, and a
**runnable example that renders the actual chart**.

**Install / import**
```
pip install matplotlib
```
```python
import matplotlib.pyplot as plt
import numpy as np
```

**The typical workflow**
1. Create data.
2. Call a plotting function like `plt.plot(...)`.
3. Add labels/titles/legend.
4. Call `plt.show()` to display (in notebooks the chart shows automatically).

## 1. Basic Plot Types
The core chart functions. Each takes your data and draws it.

### `plt.plot`

The most common function — draws a line (or points) connecting your data.

**Syntax:** `plt.plot(x, y, fmt, color=, linewidth=, linestyle=, marker=, label=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 4, 9, 16, 25])
plt.plot(x, y, color='blue', linewidth=2, marker='o')
plt.title('Line Plot')
````

### `plt.scatter`

Draws individual points — great for showing the relationship between two variables.

**Syntax:** `plt.scatter(x, y, s=, c=, marker=, alpha=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 11])
plt.scatter(x, y, c='red', s=100, alpha=0.7)
plt.title('Scatter Plot')
````

### `plt.bar`

Draws vertical bars — good for comparing categories.

**Syntax:** `plt.bar(x, height, width=0.8, color=, label=)`

````
import matplotlib.pyplot as plt
import numpy as np

categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]
plt.bar(categories, values, color='green')
plt.title('Bar Chart')
````

### `plt.barh`

Draws horizontal bars.

**Syntax:** `plt.barh(y, width, height=0.8, color=)`

````
import matplotlib.pyplot as plt
import numpy as np

categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]
plt.barh(categories, values, color='orange')
plt.title('Horizontal Bar Chart')
````

### `plt.hist`

Draws a histogram — shows how often values fall into ranges (the distribution).

**Syntax:** `plt.hist(x, bins=10, color=, edgecolor=, alpha=)`

````
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
data = np.random.randn(1000)
plt.hist(data, bins=30, color='purple', edgecolor='black')
plt.title('Histogram')
````

### `plt.pie`

Draws a pie chart — shows parts of a whole.

**Syntax:** `plt.pie(x, labels=, autopct=, colors=, explode=)`

````
import matplotlib.pyplot as plt
import numpy as np

sizes = [30, 25, 20, 25]
labels = ['Rent', 'Food', 'Travel', 'Savings']
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title('Pie Chart')
````

### `plt.boxplot`

Draws a box-and-whisker plot — summarizes spread, median, and outliers.

**Syntax:** `plt.boxplot(x, vert=True, labels=)`

````
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(1)
data = [np.random.randn(100), np.random.randn(100) + 1]
plt.boxplot(data, labels=['Group 1', 'Group 2'])
plt.title('Box Plot')
````

### `plt.stem`

Draws stems (vertical lines) from a baseline to each point.

**Syntax:** `plt.stem(x, y)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 8)
y = np.array([2, 3, 5, 4, 6, 3, 7])
plt.stem(x, y)
plt.title('Stem Plot')
````

### `plt.step`

Draws a step function (staircase line).

**Syntax:** `plt.step(x, y, where='pre')`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 8)
y = np.array([1, 3, 2, 5, 4, 6, 5])
plt.step(x, y, where='mid')
plt.title('Step Plot')
````

### `plt.fill_between`

Shades the area between two curves (or a curve and the axis).

**Syntax:** `plt.fill_between(x, y1, y2=0, color=, alpha=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.fill_between(x, y, color='skyblue', alpha=0.5)
plt.title('Fill Between')
````

### `plt.stackplot`

Draws stacked area charts — several series stacked on top of each other.

**Syntax:** `plt.stackplot(x, *ys, labels=, colors=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4]
a = [1, 2, 3, 4]; b = [1, 1, 1, 1]; c = [2, 2, 2, 2]
plt.stackplot(x, a, b, c, labels=['a', 'b', 'c'])
plt.legend()
plt.title('Stack Plot')
````

### `plt.errorbar`

Plots points/lines with error bars showing uncertainty.

**Syntax:** `plt.errorbar(x, y, yerr=, xerr=, fmt=, capsize=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 6)
y = np.array([2, 4, 6, 8, 10])
plt.errorbar(x, y, yerr=1.0, fmt='-o', capsize=5)
plt.title('Error Bar Plot')
````

### `plt.imshow`

Displays a 2-D array (or image) as a colored grid — useful for heatmaps.

**Syntax:** `plt.imshow(X, cmap=, interpolation=)`

````
import matplotlib.pyplot as plt
import numpy as np

data = np.random.rand(10, 10)
plt.imshow(data, cmap='viridis')
plt.colorbar()
plt.title('Heatmap with imshow')
````

## 2. Labels, Titles & Legends
Make a chart readable by describing what it shows.

### `plt.title`

Adds a title above the chart.

**Syntax:** `plt.title(label, fontsize=, color=, loc=)`

````
import matplotlib.pyplot as plt
import numpy as np

plt.plot([1, 2, 3], [1, 4, 9])
plt.title('My Chart Title', fontsize=14, color='navy')
````

### `plt.xlabel_ylabel`

Label the x-axis and y-axis.

**Syntax:** `plt.xlabel(text); plt.ylabel(text)`

````
import matplotlib.pyplot as plt
import numpy as np

plt.plot([1, 2, 3], [1, 4, 9])
plt.xlabel('Time (s)')
plt.ylabel('Distance (m)')
plt.title('Axis Labels')
````

### `plt.legend`

Shows a legend box. Give each plot a label= to name it.

**Syntax:** `plt.legend(loc='best', fontsize=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4]
plt.plot(x, [1, 2, 3, 4], label='Linear')
plt.plot(x, [1, 4, 9, 16], label='Quadratic')
plt.legend(loc='upper left')
plt.title('Legend Example')
````

### `plt.text`

Places text anywhere on the chart at data coordinates.

**Syntax:** `plt.text(x, y, s, fontsize=, color=)`

````
import matplotlib.pyplot as plt
import numpy as np

plt.plot([1, 2, 3], [1, 4, 9])
plt.text(1.5, 6, 'Look here!', fontsize=12, color='red')
plt.title('Text Annotation')
````

### `plt.annotate`

Adds text with an arrow pointing to a specific point.

**Syntax:** `plt.annotate(text, xy=, xytext=, arrowprops=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.annotate('peak', xy=(1.57, 1), xytext=(3, 1.2),
             arrowprops=dict(facecolor='black', width=1))
plt.title('Annotate with Arrow')
````

## 3. Axes, Limits & Grids
Control the range, ticks, and gridlines.

### `plt.xlim_ylim`

Set the visible range of each axis.

**Syntax:** `plt.xlim(min, max); plt.ylim(min, max)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.xlim(0, 5)
plt.ylim(-1.5, 1.5)
plt.title('Custom Axis Limits')
````

### `plt.xticks_yticks`

Set custom tick positions and labels.

**Syntax:** `plt.xticks(ticks, labels); plt.yticks(ticks, labels)`

````
import matplotlib.pyplot as plt
import numpy as np

plt.plot([1, 2, 3, 4], [10, 20, 25, 30])
plt.xticks([1, 2, 3, 4], ['Q1', 'Q2', 'Q3', 'Q4'])
plt.title('Custom Ticks')
````

### `plt.grid`

Adds a background grid to make values easier to read.

**Syntax:** `plt.grid(visible=True, linestyle=, alpha=)`

````
import matplotlib.pyplot as plt
import numpy as np

plt.plot([1, 2, 3, 4], [1, 4, 9, 16])
plt.grid(True, linestyle='--', alpha=0.6)
plt.title('Grid Lines')
````

### `plt.axhline_axvline`

Draw a horizontal or vertical reference line across the whole plot.

**Syntax:** `plt.axhline(y=); plt.axvline(x=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.axhline(y=0, color='red', linestyle='--')
plt.axvline(x=5, color='green', linestyle=':')
plt.title('Reference Lines')
````

### `plt.colorbar`

Adds a color scale bar next to a heatmap/scatter with colors.

**Syntax:** `plt.colorbar(mappable=None, label=)`

````
import matplotlib.pyplot as plt
import numpy as np

data = np.random.rand(8, 8)
plt.imshow(data, cmap='plasma')
plt.colorbar(label='Value')
plt.title('Colorbar')
````

## 4. Figures & Multiple Plots
Put several charts together, or control the canvas.

### `plt.figure`

Creates a new figure (the whole canvas). figsize sets width/height in inches.

**Syntax:** `plt.figure(figsize=(w, h), dpi=)`

````
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(6, 3))
plt.plot([1, 2, 3], [1, 4, 9])
plt.title('Custom Figure Size')
````

### `plt.subplot`

Adds one plot in a grid. index counts left-to-right, top-to-bottom (starts at 1).

**Syntax:** `plt.subplot(nrows, ncols, index)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 5, 50)
plt.subplot(1, 2, 1)
plt.plot(x, np.sin(x))
plt.title('sin')
plt.subplot(1, 2, 2)
plt.plot(x, np.cos(x))
plt.title('cos')
````

### `plt.subplots`

The modern way: creates a figure AND a grid of axes at once. Returns (figure, axes).

**Syntax:** `fig, ax = plt.subplots(nrows, ncols, figsize=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 5, 50)
fig, ax = plt.subplots(2, 1, figsize=(5, 4))
ax[0].plot(x, np.sin(x)); ax[0].set_title('sin')
ax[1].plot(x, np.cos(x)); ax[1].set_title('cos')
plt.tight_layout()
````

### `plt.suptitle`

Adds one big title above a whole grid of subplots.

**Syntax:** `plt.suptitle(t, fontsize=)`

````
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 5, 50)
fig, ax = plt.subplots(1, 2, figsize=(6, 3))
ax[0].plot(x, x); ax[1].plot(x, x**2)
plt.suptitle('Overall Title', fontsize=14)
````

### `plt.tight_layout`

Automatically adjusts spacing so labels/titles don't overlap.

**Syntax:** `plt.tight_layout()`

````
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(2, 2, figsize=(6, 4))
for a in ax.flat:
    a.plot([1, 2, 3], [1, 4, 9])
    a.set_title('Subplot')
plt.tight_layout()
````

## 5. Styles & Appearance

### `plt.style.use`

Applies a ready-made visual theme (e.g. 'ggplot', 'seaborn', 'dark_background').

**Syntax:** `plt.style.use(style_name)`

````
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('ggplot')
plt.plot([1, 2, 3, 4], [1, 4, 9, 16])
plt.title('ggplot Style')
plt.style.use('default')  # reset
````

## 6. Saving Your Chart
Export a figure to an image file.

### `plt.savefig`

Saves the current figure to a file (PNG, JPG, PDF, SVG...). Call it **before** `plt.show()`.

**Syntax:** `plt.savefig(fname, dpi=, bbox_inches='tight', transparent=False)`

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [1, 4, 9])
plt.title('Saved Chart')
plt.savefig('my_chart.png', dpi=150, bbox_inches='tight')
# Creates my_chart.png in your folder
```


### `plt.show`

Displays all open figures in a window. In Jupyter/Colab notebooks the chart appears
automatically, so `plt.show()` is optional there — but always include it in `.py` scripts.

**Syntax:** `plt.show()`

## Quick cheat-sheet: a complete chart

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.figure(figsize=(7, 4))              # 1. canvas
plt.plot(x, np.sin(x), label='sin(x)')  # 2. plot data
plt.plot(x, np.cos(x), label='cos(x)')
plt.title('Sine and Cosine')           # 3. describe
plt.xlabel('x'); plt.ylabel('y')
plt.legend()                            # 4. legend
plt.grid(True, alpha=0.3)
plt.savefig('chart.png', dpi=150)       # 5. save (optional)
plt.show()                              # 6. display
```
